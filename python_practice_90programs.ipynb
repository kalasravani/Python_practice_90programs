```python
# 29. Check whether a number is palindrome or not

def check_palindrome(num):
    original = num
    reverse = 0

    num = abs(num)

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num //= 10

    if original == reverse:
        return True
    else:
        return False


num = int(input("Enter a number: "))

if check_palindrome(num):
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

```python
# 34. Enter a number and print it in words

def number_to_words(num):
    words = [
        "Zero", "One", "Two", "Three", "Four",
        "Five", "Six", "Seven", "Eight", "Nine"
    ]

    if num == 0:
        return "Zero"

    if num < 0:
        return "Minus " + number_to_words(-num)

    digits = []

    while num > 0:
        digit = num % 10
        digits.append(words[digit])
        num //= 10

    digits.reverse()

    return " ".join(digits)


num = int(input("Enter a number: "))

print("Number in words =", number_to_words(num))
```

```python
# 42. Find all prime factors of a number

def prime_factors(num):
    factors = []

    divisor = 2

    while num > 1:
        if num % divisor == 0:
            factors.append(divisor)

            while num % divisor == 0:
                num //= divisor

        divisor += 1

    return factors


num = int(input("Enter a number: "))

result = prime_factors(num)

print("Prime factors are:", result)
```

```python
# 65. Count alphabets, digits and special characters in a string

def count_characters(string):
    alphabets = 0
    digits = 0
    special = 0

    for ch in string:
        if ('a' <= ch <= 'z') or ('A' <= ch <= 'Z'):
            alphabets += 1
        elif '0' <= ch <= '9':
            digits += 1
        else:
            special += 1

    return alphabets, digits, special


string = input("Enter a string: ")

alphabets, digits, special = count_characters(string)

print("Total alphabets =", alphabets)
print("Total digits =", digits)
print("Total special characters =", special)
```

```python
# 66. Count total number of vowels and consonants in a string

def count_vowels_consonants(string):
    vowels = 0
    consonants = 0

    for ch in string:
        if ('a' <= ch <= 'z') or ('A' <= ch <= 'Z'):
            if ch.lower() in "aeiou":
                vowels += 1
            else:
                consonants += 1

    return vowels, consonants


string = input("Enter a string: ")

vowels, consonants = count_vowels_consonants(string)

print("Total vowels =", vowels)
print("Total consonants =", consonants)
```

```python
# 74. Find highest frequency character in a string

def highest_frequency(string):
    frequency = {}

    for ch in string:
        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1

    highest = 0
    character = ""

    for ch in frequency:
        if frequency[ch] > highest:
            highest = frequency[ch]
            character = ch

    return character, highest


string = input("Enter a string: ")

character, frequency = highest_frequency(string)

print("Highest frequency character =", character)
print("Frequency =", frequency)
```

```python
# 75. Find lowest frequency character in a string

def lowest_frequency(string):
    frequency = {}

    for ch in string:
        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1

    lowest = min(frequency.values())
    character = ""

    for ch in frequency:
        if frequency[ch] == lowest:
            character = ch
            break

    return character, lowest


string = input("Enter a string: ")

character, frequency = lowest_frequency(string)

print("Lowest frequency character =", character)
print("Frequency =", frequency)
```

```python
# 83. Check whether a number is Prime, Armstrong or Perfect
#     using functions

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


def is_armstrong(n):
    digits = str(n)
    power = len(digits)
    total = 0

    for digit in digits:
        total += int(digit) ** power

    return total == n


def is_perfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


def check_number(n):
    if is_prime(n):
        print(n, "is a Prime number")
    else:
        print(n, "is not a Prime number")

    if is_armstrong(n):
        print(n, "is an Armstrong number")
    else:
        print(n, "is not an Armstrong number")

    if is_perfect(n):
        print(n, "is a Perfect number")
    else:
        print(n, "is not a Perfect number")


num = int(input("Enter a number: "))

check_number(num)
```

```python
# 84. Print all Prime numbers between a given interval
#     using functions

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


def print_primes(start, end):
    print("Prime numbers:")

    for num in range(start, end + 1):
        if is_prime(num):
            print(num, end=" ")


start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

print_primes(start, end)
```

```python
# 85. Print all Strong numbers between a given interval
#     using functions

def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact *= i

    return fact


def is_strong(n):
    total = 0
    temp = n

    while temp > 0:
        digit = temp % 10
        total += factorial(digit)
        temp //= 10

    return total == n


def print_strong_numbers(start, end):
    print("Strong numbers:")

    for num in range(start, end + 1):
        if is_strong(num):
            print(num, end=" ")


start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

print_strong_numbers(start, end)
```

```python
# 89. Display ATM Transaction

def atm_transaction(balance):
    while True:
        print("\n----- ATM MENU -----")
        print("1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            print("Current Balance:", balance)

        elif choice == 2:
            amount = float(input("Enter deposit amount: "))
            balance += amount
            print("Amount deposited successfully")
            print("Current Balance:", balance)

        elif choice == 3:
            amount = float(input("Enter withdrawal amount: "))

            if amount <= balance:
                balance -= amount
                print("Amount withdrawn successfully")
                print("Current Balance:", balance)
            else:
                print("Insufficient balance")

        elif choice == 4:
            print("Thank you for using ATM")
            break

        else:
            print("Invalid choice")


balance = float(input("Enter initial balance: "))

atm_transaction(balance)
```

```python
# 90. Print reverse case of a given string

def reverse_case(string):
    result = ""

    for ch in string:
        if ch.isupper():
            result += ch.lower()
        elif ch.islower():
            result += ch.upper()
        else:
            result += ch

    return result


string = input("Enter a string: ")

print("Reverse case:", reverse_case(string))
```
